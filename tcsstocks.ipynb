df.drop_duplicates(inplace=True)


**Create a new column 'Month' for grouping data by month
.dt accesses date properties, and .to_period('M') converts each date to a month-year format (e.g., 2023-01)**


```python
df['Month'] = df['Date'].dt.to_period('M')
df['Daily Return'] = df['Close'].pct_change()
```

**Create a new column 'Daily Return' to calculate the percentage change in closing price from the previous day
Formula: (Today's Close - Yesterday's Close) Yesterday's Close
.pct_change() automatically calculates this for each row**
